```python
import pandas as pd
import pylab as plt
import numpy as np
import matplotlib as mpl
```

```python
DipoleChannels=[]
DipoleChannels.append({'alpha':'d','beta':'d','alphaprime':'a','betaprime':'a'})
DipoleChannels.append({'alpha':'d','beta':'d','alphaprime':'a','betaprime':'c'})
DipoleChannels.append({'alpha':'d','beta':'d','alphaprime':'a','betaprime':'d'})
DipoleChannels.append({'alpha':'d','beta':'d','alphaprime':'c','betaprime':'c'})
DipoleChannels.append({'alpha':'d','beta':'d','alphaprime':'c','betaprime':'d'})

SpinExChannels=[]
SpinExChannels.append({'alpha':'c','beta':'c','alphaprime':'a','betaprime':'a'})
SpinExChannels.append({'alpha':'c','beta':'c','alphaprime':'a','betaprime':'c'})
SpinExChannels.append({'alpha':'c','beta':'c','alphaprime':'b','betaprime':'d'})

```

```python

n_lines = 7
cmap = mpl.colormaps['gist_heat']
# Take colors at regular intervals spanning the colormap.
colors = cmap(np.linspace(0, 1, n_lines))

n_lines = 7
cmap = mpl.colormaps['gnuplot2']
# Take colors at regular intervals spanning the colormap.
colors2 = cmap(np.linspace(0, 1, n_lines))
```

```python
dat=pd.read_csv("./Tables/Silvera_Dipoles_ref.csv")
```

```python
datH=dat[dat.Isotope=="H"]
datT=dat[dat.Isotope=="T"]
```

```python

for c in DipoleChannels:
    plt.figure(figsize=(5,5),dpi=250)

    datCH=datH[datH.Channel==str(c)]
    i=0
    for T in datCH[('T')].unique():
        datCHT=datCH[datCH[('T')]==T]
        plt.plot(datCHT.B.values,datCHT.sigma.values,label=T,color=colors[i])
        i=i+1
    plt.legend(ncol=3,title="Temperature (K)")
    Title=r"Hydrogen, "+ c['alpha']+c['beta']+r"$\rightarrow$"+c['alphaprime']+c['betaprime']
    plt.title(Title)
    plt.loglog()
    plt.ylabel(r"$\sigma$ (a.u.)$^2$")
    plt.xlabel(r"B(T)")

    plt.ylim(1e-8,1)
    plt.xlim(1e-3,10)
    plt.grid(which='both')
    fname=r"Hydrogen_sigma_"+ c['alpha']+c['beta']+r"_to_"+c['alphaprime']+c['betaprime']+".png"
    plt.savefig("Plots/"+fname,dpi=250,bbox_inches='tight')
    plt.show()







```

```python
for c in DipoleChannels:
    plt.figure(figsize=(5,5),dpi=250)


    datCT=datT[datT.Channel==str(c)]
    i=0
    for T in datCT[('T')].unique():
        datCTT=datCT[datCT[('T')]==T]
        plt.plot(datCTT.B.values,datCTT.sigma.values,'-',label=T,color=colors2[i])
        i=i+1
    plt.legend(ncol=3,title="Temperature (K)")
    Title=r"Tritium, "+ c['alpha']+c['beta']+r"$\rightarrow$"+c['alphaprime']+c['betaprime']
    plt.title(Title)
    plt.ylabel(r"$\sigma$ (a.u.)$^2$")
    plt.xlabel(r"B(T)")
    plt.loglog()
    plt.ylim(1e-5,1000)
    plt.xlim(1e-3,10)
    plt.grid(which='both')
    fname=r"Tritium_sigma_"+ c['alpha']+c['beta']+r"_to_"+c['alphaprime']+c['betaprime']+".png"
    plt.savefig("Plots/"+fname,dpi=250,bbox_inches='tight')
    plt.show()
```

```python
plt.figure(figsize=(5,5),dpi=250)
i=0
for T in datsT[('T')].unique():
    datsTT=datsT[datsT[('T')]==T]
    G=[]
    for c in SpinExChannels:
        datsTTC=datsTT[datsTT.Channel==str(c)]
        if(len(G)==0):
            G=datsTTC.G.values
        else:
            G=G+datsTTC.G.values
    plt.plot(datsTTC.B.values,G,label=T,color=colors2[i])
    i=i+1
plt.legend(ncol=2,title="Temperature (K)")
Title=r"Tritium, cc$\rightarrow$all"
plt.title(Title)
plt.loglog()
plt.xlabel(r"B (T)")
plt.ylabel(r"G (cm$^3$s$^{-1}$)")

plt.ylim(1e-14,1e-9)
plt.xlim(1e-3,10)
plt.grid(which='both')
fname=r"Tritium_G_cc_to_all.png"
plt.savefig("Plots/"+fname,dpi=250,bbox_inches='tight')
plt.show()
```
